```python
import numpy as np
import pandas as pd
import re

seed= 42
```

```python
train_df= pd.read_csv('train.csv')
test_df= pd.read_csv('test.csv')

X_train= train_df['text']
y_train= train_df['label']

X_test= test_df['text']
```

```python
# Subtask 1

answer1= pd.DataFrame([{
    'subtaskID': 1,
    'datapointID': id_,
    'answer': len(text)
}for id_, text  in zip(test_df['sample_id'],test_df['text'])])

answer1.head()
```

```python
# Subtask 2

answer2= pd.DataFrame([{
    'subtaskID': 2,
    'datapointID': id_,
    'answer': len(re.findall(r'\bfree\b', text, flags=re.IGNORECASE))
}for id_, text  in zip(test_df['sample_id'],test_df['text'])])

answer2.head()
```

```python
# Subtask 3

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.svm import LinearSVC

cv= CountVectorizer()

X_train= cv.fit_transform(X_train)
X_test= cv.transform(X_test)

model= LinearSVC(max_iter=1000)

model.fit(X_train, y_train)

predictions= model.predict(X_test)
```

```python
answer3= pd.DataFrame([{
    'subtaskID': 3,
    'datapointID': id_,
    'answer': pred
}for id_, pred  in zip(test_df['sample_id'], predictions)])

answer3.head()
```

```python
answer= pd.concat([answer1, answer2, answer3])
answer.to_csv("submission.csv", index= False)
```
