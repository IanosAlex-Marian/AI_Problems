```python
import numpy as np
import pandas as pd
```

```python
train_df= pd.read_csv('muzee.csv')
```

```python
# Subtask 1
answer1= pd.DataFrame([{
    "id": 1,
    "subtaskID": 1,
    "answer": len(train_df)
}])

answer1.head()
```

```python
# Subtask 2
answer2= pd.DataFrame([{
    "id": 2,
    "subtaskID": 2,
    "answer": len(train_df[train_df['județul'] == 'București'])
}])

answer2.head()
```

```python
# Subtask 3
answer3= pd.DataFrame([{
    "id": 3,
    "subtaskID": 3,
    "answer": np.sum(train_df.isna().sum().values != 0)
}])

answer3.head()
```

```python
# Subtask 4
answer4= pd.DataFrame([{
    "id": 4,
    "subtaskID": 4,
    "answer": train_df['anul înființării'].value_counts().index[0]
}])

answer4.head()
```

```python
# Subtask 5
answer5= pd.DataFrame([{
    "id": index,
    "subtaskID": 5,
    "answer": train_df['județul'].value_counts().loc[index]
} for index in train_df['județul'].value_counts().index])

answer5.head()
```

```python
# Subtask 6
answer6= pd.DataFrame([{
    "id": train_df.iloc[idx]["_id"],
    "subtaskID": 6,
    "answer": round(train_df.iloc[idx].notna().sum()/ len(train_df.columns)* 100, 2)
} for idx in train_df.index])

answer6.head()
```

```python
# Subtask 7
answer7= pd.DataFrame([{
    "id": 7,
    "subtaskID": 7,
    "answer": round(answer6['answer'].sum()/len(train_df), 2)
} ])

answer7.head()
```

```python
# Subtask 8
answer8= pd.DataFrame([{
    "id": 8,
    "subtaskID": 8,
    "answer": round(len(answer6[answer6['answer'] == answer6['answer'].max()])/ len(answer6) * 100, 2)
} ])

answer8.head()
```

```python
answer= pd.concat([answer1, answer2, answer3, answer4, answer5, answer6, answer7, answer8])
answer.to_csv("submission.csv", index= False)
```
